# College Predictor - data preparation
Loads `colleges.json`, explores cutoffs and shows the same prediction logic used by the web page.

```python
import pandas as pd
df = pd.read_json('colleges.json')
df.sort_values('cutoff', ascending=False)
```

```python
df.groupby('branch')['cutoff'].describe()
```

```python
def classify(p, cutoff):
    d = p - cutoff
    if d >= 3: return 'safe'
    if d >= 0: return 'target'
    if d >= -5: return 'reach'
    return None

percentile = 92.5
df['chance'] = df['cutoff'].apply(lambda c: classify(percentile, c))
df.dropna(subset=['chance'])
```
